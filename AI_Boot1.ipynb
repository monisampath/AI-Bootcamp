```python
import json

# 1. Initialize conversation history with the system prompt outside the function
conversation_history = [
    {
        "role": "system",
        "content": "Act as a professional nutritionist. Keep track of user details like their name, dietary goals, and preferences shared during the conversation."
    }
]

def nutrition_agent(user_prompt):
    # 2. Append the user's new message to the history
    conversation_history.append({
        "role": "user",
        "content": user_prompt
    })

    # 3. Pass the entire conversation_history to the model
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=conversation_history,
        temperature=0.3
    )

    assistant_reply = response.choices[0].message.content

    # 4. Append the assistant's reply to the history so it remembers what it said
    conversation_history.append({
        "role": "assistant",
        "content": assistant_reply
    })

    return assistant_reply
```

```python
# Turn 1: Give it context
print("Bot:", nutrition_agent("Hi! My name is Alex and I want to build muscle on a vegetarian diet."))
print("-" * 50)

# Turn 2: Ask it to recall the context
print("Bot:", nutrition_agent("Tell me my name and suggest a high-protein breakfast for my goal."))
```
